```python
import spacy
```

### Using continual learning ner combined

```python
# Entity classifier
medical_nlp = spacy.load("continual_learning_ner_combined")
```

### Using med 7

```python
# Sentence segmentation
nlp = spacy.load("en_core_web_sm")
```

### Using sci spacy models -- sci md

#### Run test samples with whichever model is currently loaded -- Test samples

```python
texts = [
            "I have a stomachache",
            "I feel dizzy",
            "I have teary eyes and a runny nose",
            "I have a headache and a sore throat",
            "I have a fever and a cough"
        ]

index = 1
for index in range(len(texts)):
    doc = nlp(texts[index])
    med_doc = medical_nlp(texts[index])
    entities = dict()

    for ent in med_doc.ents:
        print(f" {ent.label_} : {ent.text} ")
        entities[ent.label_] = ent.text
    for ent in doc.ents:
        print(f" {ent.label_} : {ent.text} \n")
        entities[ent.label_] = ent.text
    print("The final output >> ", entities, "\n---------------------------------------------")
```
