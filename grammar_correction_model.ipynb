### Here we test how to correct incorrect input from the voice transcription models and pick the best method
* OCR Will use a different method since it can get words not in a recognized language
* We shall use vosk for simulations of incorrect and incoherent VTM(Voice Transcription Model) outputs

```python
from happytransformer import HappyTextToText, TTSettings

happy_tt = HappyTextToText("T5", "vennify/t5-base-grammar-correction")

args = TTSettings(num_beams=5, min_length=1)

# Add the prefix "grammar: " before each input 
result = happy_tt.generate_text("grammar: This sentences has has bads grammar.", args=args)

print(result.text) # This sentence has bad grammar.

```
